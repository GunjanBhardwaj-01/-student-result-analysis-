```python
import pandas as pd
import io

csv_data = """Name,Maths,Science,English
Aman,80,75,85
Riya,90,88,92
Rahul,60,65,70
Sneha,95,92,96
Karan,50,55,60"""

df = pd.read_csv(io.StringIO(csv_data))
print(df)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = df

print("First 5 rows:")
print(data.head())
```

```python
data['Total'] = data[['Maths','Science','English']].sum(axis=1)
data['Average'] = data['Total'] / 3

print(data)
```

```python
topper = data.loc[data['Total'].idxmax()]
print("Topper:\n", topper)
```

```python
data['Result'] = data['Average'].apply(lambda x: 'Pass' if x >= 40 else 'Fail')

print(data[['Name','Average','Result']])
```

```python
plt.figure()
plt.bar(data['Name'], data['Total'])
plt.title("Total Marks of Students")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
plt.figure()
plt.plot(data['Name'], data['Average'])
plt.title("Average Marks")
plt.xlabel("Students")
plt.ylabel("Average")
plt.show()
```

```python
subject_avg = data[['Maths','Science','English']].mean()

plt.figure()
subject_avg.plot(kind='bar')
plt.title("Subject-wise Average")
plt.show()
```
